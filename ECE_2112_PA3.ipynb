```python
import pandas as pd

new_cars = pd.read_csv('cars.csv')
```

## **A. Positional and Label-Based Slicing**

```python
df_1 = pd.DataFrame(new_cars)
print("Shape of Dataframe:", df_1.shape)
display(df_1)


```

```python
cars_6_to_10 = df_1.iloc[5:10]
selected_columns = cars_6_to_10.loc[:, ['Model', 'mpg', 'cyl','hp', 'gear']]
print("Cars 6 to 10 (Model up to Gear):", "\n")
selected_columns
```

## **B. Model Lookup**

```python
df_2 = pd.DataFrame(new_cars)
toyota = df_2.loc[new_cars['Model'] == "Toyota Corolla"]
display(toyota)

pontiac = df_2.loc[new_cars['Model'] == 'Pontiac Firebird', ['Model', 'mpg', 'hp', 'wt']]
display(pontiac)
```

## **C. Multi-Model Subsetting**

```python
models_to_select = ['Datsun 710', 'Lotus Europa', 'Ferrari Dino']
columns_to_keep = ['Model', 'mpg', 'cyl', 'hp', 'gear']

selected_cars = new_cars[new_cars['Model'].isin(models_to_select)][columns_to_keep]

display(selected_cars)
print("\n","Shape of Selected Cars Dataframe:", selected_cars.shape)
```
